```python
import turtle

# Draw a square
for _ in range(4):
    turtle.forward(100)
    turtle.right(90)

turtle.done()
```

```python
import turtle

turtle.speed(0)  # Fastest animation
turtle.colormode(255)

for i in range(100):
    turtle.pencolor((i*2 % 255, i*3 % 255, i*5 % 255))
    turtle.forward(i * 2)
    turtle.left(91)

turtle.done()
```
