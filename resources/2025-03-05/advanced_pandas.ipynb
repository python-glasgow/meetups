## Memory efficiency using specific dtypes

```python
import pandas as pd

# Create a DataFrame from a dictionary with three columns, first column is an integer, second column is a float, and third column is a string. 
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4.3, 5.7, 6.2],
    'C': ['red', 'red', 'blue']
})
df
```

```python
df.info(memory_usage='deep')
```

```python
from random import randint, random, choice
C = ["red", "blue", "green"]

rows = []
for i in range(1_000_000):
    
    a = randint(1, 100)        
    b = randint(1, 100) + random()  
    c = choice(C)
    
    rows.append([a, b, c])

df = pd.DataFrame(rows, columns=["A", "B", "C"])
df      
    
```

```python
df.info(memory_usage='deep')
```

## Chain functions using pipe

```python
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4.3, 5.7, 6.2],
    'C': ['red', 'red', 'blue']
})
print(df)
```

```python
# Multiply 'A' by 10
df['A'] = df['A'] * 10

# Round 'B'
df['B'] = df['B'].round()

# Filter out rows where 'C' is 'red'
df_filtered = df[df['C'] != 'blue']

print(df_filtered)
```

```python
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4.3, 5.7, 6.2],
    'C': ['red', 'red', 'blue']
})
# Define our transformation functions
def multiply_A(df):
    df['A'] = df['A'] * 10
    return df

def round_B(df):
    df['B'] = df['B'].round()
    return df

def filter_blue(df):
    return df[df['C'] != 'blue']

# Chain the functions using pipe
df_result = (
    df
    .pipe(multiply_A)
    .pipe(round_B)
    .pipe(filter_blue)
)

print(df_result)
```

Cleaner code, better readability, modularity and reusability.

## df.query 

```python
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4.3, 5.7, 6.2],
    'C': ['red', 'red', 'blue']
})
df
```

## Joining dfs using merge, SQL style

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'A': [1, 2, 3],
    'B': [4.3, 5.7, 6.2],
    'C': ['red', 'red', 'blue']
})


df2 = pd.DataFrame({
    'ID': [ 2, 3, 4, 5],
    'D': ['Berry', 'Cherry', 'Orange', 'Banana'],
})
```

```python
inner_joined = pd.merge(df1, df2, on='ID', how='inner')
print("Inner Join Result:")
print(inner_joined)
```

```python
outer_joined = pd.merge(df1, df2, on='ID', how='outer')
print("Outer Join Result:")
print(outer_joined)
```

##  MultiIndex

```python
df = pd.DataFrame({
    'Store': ['Store A', 'Store A', 'Store A', 'Store B', 'Store B', 'Store B'],
    'Department': ['Electronics', 'Clothing', 'Grocery', 'Electronics', 'Clothing', 'Grocery'],
    'Sales': [20000, 15000, 30000, 25000, 12000, 28000],
    'Profit': [5000, 3000, 7000, 6000, 2500, 6500]
})

# Create a MultiIndex by setting 'Store' and 'Department' as the index
df_multi = df.set_index(['Store', 'Department'])
print(df_multi)
```

```python
clothing_data = df_multi.xs('Store A', level='Store')
print(clothing_data)
```

```python
clothing_data = df_multi.xs('Clothing', level='Department')
print(clothing_data)
```

```python
store_b_clothing = df_multi.loc[('Store B', 'Clothing')]
print(store_b_clothing)
```

```python
total_sales_by_store = df_multi['Sales'].groupby(level='Store').sum()
print(total_sales_by_store)
```

```python
total_sales_by_department = df_multi['Sales'].groupby(level='Department').sum()
print(total_sales_by_department)
```

##  Reshaping

- Long to wide

```python
df = pd.DataFrame({
    'Store': ['Store A', 'Store A', 'Store B', 'Store B', 'Store A', 'Store B'],
    'Department': ['Electronics', 'Clothing', 'Electronics', 'Clothing', 'Electronics', 'Clothing'],
    'Month': ['Jan', 'Jan', 'Jan', 'Jan', 'Feb', 'Feb'],
    'Sales': [20000, 15000, 25000, 12000, 21000, 13000]
})

print("Original DataFrame:")
print(df)
```

```python
df_pivot = df.pivot(index=['Store', 'Month'], columns='Department', values='Sales')
print("\nPivoted DataFrame (Wide Format):")
print(df_pivot)
```

- Wide to long

```python
df_reset = df_pivot.reset_index()

df_melted = pd.melt(df_reset, id_vars=['Store', 'Month'], 
                    var_name='Department', value_name='Sales')

print("\nMelted DataFrame (Long Format):")
print(df_melted)
```

- Stacking

```python
print(df_pivot)
```

```python
df_stacked = df_pivot.stack()
print(df_stacked)
```

```python
df_unstacked = df_stacked.unstack()
print(df_unstacked)
```

```python
df_unstacked = df_stacked.unstack().unstack().unstack().unstack()
print(df_unstacked)
```

##  Grouping and aggregation

```python
df = pd.DataFrame({
    'Store': ['Store A', 'Store A', 'Store B', 'Store B', 'Store A', 'Store B'],
    'Department': ['Electronics', 'Clothing', 'Electronics', 'Clothing', 'Electronics', 'Grocery'],
    'Sales': [20000, 15000, 25000, 12000, 21000, 13000],
    'Profit': [5000, 3000, 6000, 2500, 7000, 3500]
})

df_multi = df.set_index(['Store', 'Department'])
print(df_multi)
```

```python
store_group = df_multi.groupby(level='Store').agg({
    'Sales': 'sum',
    'Profit': 'mean'
})
print(store_group)
```

```python
custom_agg = df_multi.groupby(level='Store').agg({
    'Sales': [
        'sum',                # Total sales
        'mean',               # Average sales
        lambda x: x.max() - x.min()  # Sales range
    ],
    'Profit': [
        'sum',                # Total profit
        'mean',               # Average profit
        lambda x: x.std()     # Profit standard deviation
    ]
})

print(custom_agg)
```

## Window functions

- Moving average

```python
dates = pd.date_range('2025-01-01', periods=10, freq='D')
df = pd.DataFrame({
    'Date': dates,
    'Sales': [200, 220, 250, 270, 260, 300, 320, 310, 330, 340]
})
df.set_index('Date', inplace=True)
print(df)
```

```python
moving_avg_manual = []
sales_values = df['Sales'].tolist()
window_size = 3

for i in range(len(sales_values)):
    if i < window_size - 1:
        moving_avg_manual.append(None)
    else:
        window_sum = sum(sales_values[i-window_size+1:i+1])
        avg = window_sum / window_size
        moving_avg_manual.append(avg)

df['Moving_Avg_Manual'] = moving_avg_manual

print(df)
```

```python
df['Moving_Avg_Window'] = df['Sales'].rolling(window=3).mean()
print(df)
```
